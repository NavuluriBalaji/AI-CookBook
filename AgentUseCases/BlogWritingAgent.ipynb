### 💡 **nbagents Makes This Simple:**

agent = Agent(llm=your_llm_function)

````python
from nbagents import Agent
from google.colab import userdata
import google.generativeai as genai
import requests
from bs4 import BeautifulSoup
import json
from groq import Groq

def call_llm_gemini(prompt: str) -> str:
    """Use Gemini for blog creation and content generation tasks"""
    api_key = userdata.get("GOOGLE_API_KEY")
    genai.configure(api_key=api_key)
    model = genai.GenerativeModel('gemini-2.5-pro')

    response = model.generate_content(prompt)
    return response.text

def call_llm_grok(prompt: str) -> str:
    """Use Grok for research and question generation tasks"""
    try:

        import os
        os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")

        client = Groq()
        completion = client.chat.completions.create(
            model="qwen/qwen3-32b",  # Using a reliable Groq model. Change for your use case Accordingly
            messages=[
                {
                    "role": "system",
                    "content": "You are a helpful AI assistant specialized in research and information gathering. Provide comprehensive, well-structured responses."
                },
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            temperature=0.7,
            max_tokens=8192,
            top_p=1,
            stream=False
        )

        return completion.choices[0].message.content

    except Exception as e:
        print(f"Groq API error: {str(e)}")
        print("Falling back to Gemini...")
        # Fallback to Gemini if Groq fails
        return call_llm_gemini(prompt)


agent = Agent(llm=call_llm_gemini)

@agent.tool("generate_research_questions")
def generate_research_questions(topic: str) -> str:
    """Generate comprehensive research questions using Groq API (optimized for research tasks)"""
    question_prompt = f"""
    Generate 15-20 comprehensive research questions about "{topic}" to gather complete information for a blog post.

    Include questions covering:
    - WHAT: Basic definitions, concepts, key aspects
    - WHY: Importance, benefits, reasons, motivations
    - HOW: Methods, processes, step-by-step guides
    - WHEN: Timeline, history, current trends, future predictions
    - WHERE: Locations, applications, markets, demographics
    - WHO: Key players, experts, target audience, stakeholders
    - WHICH: Comparisons, alternatives, best practices
    - STATISTICS: Latest data, numbers, market size, growth rates
    - TRENDS: Current developments, future outlook, emerging patterns
    - CHALLENGES: Problems, obstacles, limitations, solutions

    Format as a numbered list of specific, searchable questions.
    Make questions specific and detailed for better search results.

    Example format:
    1. What is [specific aspect of topic]?
    2. How has [topic] evolved in the last 5 years?
    3. What are the latest statistics on [topic] in 2024-2025?
    """
    return call_llm_grok(question_prompt)

@agent.tool("analyze_research_data")
def analyze_research_data(research_data: str, topic: str) -> str:
    """Analyze and summarize research data using Groq API (optimized for data processing)"""
    analysis_prompt = f"""
    Analyze this comprehensive research data about "{topic}" and create a well-organized summary:

    RESEARCH DATA:
    {research_data}

    Please provide:
    1. KEY FINDINGS: Most important insights and statistics
    2. CURRENT TRENDS: What's happening now in 2024-2025
    3. MAIN THEMES: Common patterns across the research
    4. ACTIONABLE INSIGHTS: Practical takeaways
    5. DATA POINTS: Important statistics and numbers
    6. EXPERT OPINIONS: Notable quotes or perspectives
    7. FUTURE PREDICTIONS: What's expected to happen

    Organize this information clearly and concisely for blog writing.
    """

    return call_llm_grok(analysis_prompt)

@agent.tool("search_web")
def search_web(query: str, num_results: int = 10) -> str:
    """Search the web using Serper API for information about a topic. Returns search results."""
    try:
        serper_api_key = userdata.get("SERPER_API_KEY")

        url = "https://google.serper.dev/search"

        payload = json.dumps({
            "q": query,
            "num": num_results
        })

        headers = {
            'X-API-KEY': serper_api_key,
            'Content-Type': 'application/json'
        }

        response = requests.request("POST", url, headers=headers, data=payload)
        response.raise_for_status()

        results = response.json()

        formatted_results = f"Search results for '{query}':\n\n"

        if 'organic' in results:
            for i, result in enumerate(results['organic'][:num_results], 1):
                title = result.get('title', 'No title')
                link = result.get('link', 'No link')
                snippet = result.get('snippet', 'No description')

                formatted_results += f"{i}. {title}\n"
                formatted_results += f"   URL: {link}\n"
                formatted_results += f"   Description: {snippet}\n\n"

        if 'knowledgeGraph' in results:
            kg = results['knowledgeGraph']
            formatted_results += f"Knowledge Graph:\n"
            formatted_results += f"Title: {kg.get('title', 'N/A')}\n"
            formatted_results += f"Description: {kg.get('description', 'N/A')}\n\n"

        if 'answerBox' in results:
            ab = results['answerBox']
            formatted_results += f"Answer Box:\n"
            formatted_results += f"Answer: {ab.get('answer', ab.get('snippet', 'N/A'))}\n\n"

        return formatted_results

    except Exception as e:
        return f"Error searching web: {str(e)}\nMake sure you have set your SERPER_API_KEY in userdata."

@agent.tool("research_multiple_queries")
def research_multiple_queries(questions_list: str) -> str:
    """Take a list of questions and search for each one, returning comprehensive research data."""
    try:
        questions = []
        lines = questions_list.split('\n')
        for line in lines:
            line = line.strip()
            if line and (line[0].isdigit() or line.startswith('-')):
                question = line.split('.', 1)[-1].strip() if '.' in line else line[1:].strip()
                if len(question) > 10:
                    questions.append(question)

        all_research = ""

        for i, question in enumerate(questions[:8], 1):  # Reduced to 8 questions to manage token limits
            search_results = search_web(question, 4)  # 4 results per question
            all_research += f"\n{'='*60}\nRESEARCH QUESTION {i}: {question}\n{'='*60}\n"
            all_research += search_results + "\n"

        return all_research

    except Exception as e:
        return f"Error in research: {str(e)}"

@agent.tool("create_medium_blog")
def create_medium_blog(topic: str, research_summary: str, target_audience: str = "general readers") -> str:
    """Create a beautifully formatted Medium-style blog post using Gemini API (optimized for creative writing)"""

    blog_prompt = f"""
    Create a high-quality, engaging Medium-style blog post about "{topic}" for {target_audience}.

    Use this research summary: {research_summary[:6000]}  # Limit research data size for Gemini

    BLOG REQUIREMENTS:

    1. TITLE: Create a compelling, click-worthy title (use numbers, power words, emotional triggers)
    2. SUBTITLE: Add an engaging subtitle that expands on the title
    3. INTRODUCTION: Start with a hook (story, statistic, question, or bold statement)
    4. STRUCTURE: Use clear sections with engaging subheadings
    5. CONTENT: Write in a conversational, engaging tone with:
       - Real data and statistics from the research
       - Practical insights and actionable advice
       - Stories or examples to illustrate points
       - Current trends and future predictions
    6. FORMATTING: Use Medium-style formatting:
       - Short paragraphs (1-3 sentences)
       - Bullet points and numbered lists
       - Bold key phrases
       - Engaging subheadings
    7. CONCLUSION: End with key takeaways and a thought-provoking question
    8. CALL-TO-ACTION: Invite readers to engage (comment, share, follow)

    WRITING STYLE:
    - Conversational and accessible
    - Data-driven but not dry
    - Include current statistics and trends
    - Use storytelling elements
    - Make it scannable and easy to read

    LENGTH: Aim for 1200-1800 words for optimal Medium engagement.

    Make it feel like a professional Medium article that would get thousands of claps!
    """

    return call_llm_gemini(blog_prompt)

@agent.tool("scrape_website")
def scrape_website(url: str) -> str:
    """Scrape content from a website URL and return the text content."""
    try:
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
        }
        response = requests.get(url, headers=headers, timeout=10)
        response.raise_for_status()

        soup = BeautifulSoup(response.content, 'html.parser')

        for script in soup(["script", "style"]):
            script.extract()

        text = soup.get_text()

        lines = (line.strip() for line in text.splitlines())
        chunks = (phrase.strip() for line in lines for phrase in line.split("  "))
        text = ' '.join(chunk for chunk in chunks if chunk)

        return text[:3000]
    except Exception as e:
        return f"Error scraping website: {str(e)}"

@agent.tool("generate_image_prompt")
def generate_image_prompt(blog_topic: str, style: str = "professional") -> str:
    """Generate a detailed image prompt using Gemini (optimized for creative tasks)"""
    prompt_creation = f"""
    Create a detailed, creative image generation prompt for a Medium blog post about "{blog_topic}".

    Style: {style}

    The prompt should include:
    - Visual elements related to the topic
    - Color schemes and mood
    - Composition and layout suggestions
    - Typography or text elements
    - Professional quality indicators
    - Social media optimization aspects

    Make it detailed enough for AI image generators like DALL-E, Midjourney, or Stable Diffusion.
    """

    return call_llm_gemini(prompt_creation)
````
